# SV2 — tuần 5 và 6
Chỉ đọc hồ sơ Git và kiểm checksum. Không build/chạy model. Kernel PC có Python chuẩn là đủ.
Khớp lớp FP32 không phải accuracy độc lập; root timing không phải DPU-only.


```python
from pathlib import Path
import json, subprocess
from IPython.display import display, HTML
ROOT = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p/'edge/runtime').is_dir())
for week in (5,6):
    folder=ROOT/f'results/week{week}/edge'
    subprocess.run(['sha256sum','-c','sha256sum_export.txt'],cwd=folder,check=True)
print('Nguồn:',ROOT)
```

```python
w5=json.loads((ROOT/'results/week5/edge/summary.json').read_text())
print('Tuần 5:',[(r['split'],r['samples'],r['max_abs_error'],r['class_agreement']) for r in w5])
p=json.loads((ROOT/'results/week6/edge/profile_edge.json').read_text())
print('Tuần 6:',p['created_utc'],p['status'],p['configuration'])
rows=[]
for r in p['splits']:
    rows.append('<tr>'+''.join('<td>'+str(x)+'</td>' for x in [r['split'],r['loading']['model_load_ms']['mean_ms'],r['steady']['execute_wait_ms']['mean_ms'],r['steady']['tail_total_ms']['mean_ms'],r['steady']['tail_total_ms']['p95_ms']])+'</tr>')
display(HTML('<table><tr><th>s</th><th>Load mean ms</th><th>Root execute mean ms</th><th>Tail mean ms</th><th>Tail p95 ms</th></tr>'+''.join(rows)+'</table>'))
print('DPU-only:',p['dpu_only_timing'],'| accuracy độc lập:',p['independent_test_accuracy'])
```
